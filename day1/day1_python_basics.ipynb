## Variables and types

Text is a string, whole numbers are ints, decimals are floats. An f-string drops a value
straight into text.

```python
party = 'Labour'
seats = 412
share = 0.34
print(f'{party} won {seats} seats on {share:.0%} of the vote')
```

## Lists, an ordered collection

A list is R's vector. You index from zero, slice with a colon in the brackets, and loop over it.

```python
parties = ['left', 'right', 'left', 'right', 'left']
print('first', parties[0])
print('first two', parties[:2])
parties.append('right')
print('length', len(parties))
for p in parties:
    print('saw', p)
```

## List comprehensions, the sapply of Python

A comprehension builds a new list from an old one in one line.

```python
lengths = [len(p) for p in parties]
lefties = [p for p in parties if p == 'left']
print(lengths)
print(lefties)
```

## Dictionaries, the label maps you will use everywhere

A dictionary maps a key to a value. Label-to-id maps and config live in dictionaries, so this
object matters more downstream than any other.

```python
label_to_id = {'left': 0, 'right': 1}
print('left is', label_to_id['left'])
print('keys', list(label_to_id.keys()))
for name, i in label_to_id.items():
    print(name, '->', i)
```

## pandas, your data.frame in Python

pandas is the data.frame you know. Load the small labeled file that ships with this notebook,

## Inspect and filter

`value_counts` tabulates a column, a boolean mask filters rows, and `groupby` summarizes.

```python
print(df['label'].value_counts())
print('--- one filtered row ---')
df[df['label'] == 'left'].head(2)
```

## Clean the messy table

Real files are messy. This one has stray spaces and labels in mixed case. Strip the text,
lowercase the labels, drop blanks, then build the label-to-id map the models will need.

```python
df['text'] = df['text'].str.strip()
df['label'] = df['label'].str.strip().str.lower()
df = df.dropna(subset=['text', 'label']).reset_index(drop=True)

labels = sorted(df['label'].unique())
label_to_id = {name: i for i, name in enumerate(labels)}
df['label_id'] = df['label'].map(label_to_id)
print('label_to_id', label_to_id)
print(df['label'].value_counts())
df.head()
```

## Reading a traceback

Errors are normal. Read the last line first, it names what went wrong. Here we ask for a
column that does not exist and catch the error so the notebook keeps running.

```python
try:
    df['patry']            # a typo for 'party' that is not even in this file
except KeyError as e:
    print('KeyError:', e)
    print('Read the last line. A KeyError means the column name is wrong or missing.')
```
